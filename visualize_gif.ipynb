```python
from PIL import Image
import matplotlib.pyplot as plt
from ipywidgets import interact, IntSlider, FloatSlider

gif = Image.open("test_result/hybrid/visualizations/state_seed_46.gif")

frames = []

for i in range(gif.n_frames):
    gif.seek(i)
    frames.append(gif.convert("RGB").copy())


def show_frame(frame, zoom):
    fig, ax = plt.subplots(figsize=(10 * zoom, 8 * zoom), dpi=120)

    ax.imshow(frames[frame], interpolation="nearest")
    ax.set_title(f"Frame {frame + 1}/{len(frames)}", fontsize=16)
    ax.axis("off")

    plt.show()


interact(
    show_frame,
    frame=IntSlider(
        min=0,
        max=len(frames) - 1,
        step=1,
        value=0,
        description="Frame"
    ),
    zoom=FloatSlider(
        min=0.5,
        max=2.5,
        step=0.25,
        value=1.5,
        description="Zoom"
    )
);
```
